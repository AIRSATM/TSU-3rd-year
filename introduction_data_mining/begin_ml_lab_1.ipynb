#Разведочный анализ данных о бриллиантах
Objectives:
Данный проект представляет собой комплексный разведочный анализ данных (Exploratory Data Analysis, EDA) для набора данных diamonds, содержащего информацию о характеристиках и ценах бриллиантов.

DataSet: diamonds из библиотеки Seaborn

Описание: Встроенный датасет содержит 53,940 наблюдений и 10 признаков, описывающих физические свойства бриллиантов (вес, размеры, качество огранки, цвет, чистота) и их рыночную стоимость.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

df = sns.load_dataset('diamonds')

display(df.head())
```

```
df.info()

print(f"\nразмер датасета: {df.shape[0]} строк, {df.shape[1]} столбцов")


missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])

duplicates = df.duplicated().sum()

if duplicates > 0:
    df = df.drop_duplicates()
    print(f"дубликаты удалены. новый размер: {df.shape}")
```

```

numeric_df = df.select_dtypes(include=[np.number])

corr_matrix = numeric_df.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('матрица корреляции числовых признаков')
plt.show()
```

```
plt.figure(figsize=(10, 6))
sns.histplot(df['price'], bins=30, kde=True, color='blue')
plt.title('Распределение цен на бриллианты')
plt.xlabel('Цена ($)')
plt.ylabel('Количество')
plt.show()
```

```
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='carat', y='price', alpha=0.5)
plt.title('Зависимость цены от веса (Carat vs Price)')
plt.xlabel('Вес (Carat)')
plt.ylabel('Цена')
plt.show()
```

```
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='cut', y='price', palette='Set2')
plt.title('Распределение цен по качеству огранки')
plt.show()
```

```
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='color', hue='cut', palette='viridis')
plt.title('Количество бриллиантов по цвету и огранке')
plt.show()
```

```
pivot_table = df.pivot_table(index='color', columns='clarity', values='price', aggfunc='mean')

plt.figure(figsize=(10, 6))
sns.heatmap(pivot_table, cmap='YlGnBu', annot=True, fmt=".0f")
plt.title('Средняя цена: Цвет vs Прозрачность')
plt.show()
```

#Conclusion

Data structure: Набор данных чистый, пропусков нет. Было обнаружено небольшое количество дубликатов, которые были удалены. Типы данных соответствуют ожидаемым (числовые для размеров и цены, категориальные для характеристик качества).

Correlation: Самая сильная прямая корреляция наблюдается между весом (carat) и ценой (price), а также между весом и физическими размерами (x, y, z). Это логично: чем крупнее камень, тем он дороже.

Distribution: Гистограмма цен показывает "длинный хвост" справа. Это значит, что большинство бриллиантов стоят недорого, но есть редкие экземпляры с очень высокой ценой.

Categorical features: График "ящик с усами" (Boxplot) показал контринтуитивный результат: медианная цена бриллиантов с огранкой "Fair" (плохая) иногда выше, чем у "Ideal". Это может быть связано с тем, что камни "Fair" часто имеют больший вес в каратах, что перекрывает недостаток огранки.
